**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Assessing Data

#### Identifying ... problem

```python
# Mengecek missing value
missing_data = df.isna().sum()
missing_data = missing_data[missing_data > 0]

print("Missing value:")
print(missing_data)

# Mengecek jumlah outlier dengan metode IQR
kolom_polutan = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3']

print("\nJumlah outlier:")

for kolom in kolom_polutan:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    jumlah_outlier = (
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ).sum()

    print(f"{kolom}: {jumlah_outlier}")
```

**Steps to Take:**

- Outlier akan diperiksa lebih lanjut dan tidak langsung dihapus karena nilai ekstrem pada konsentrasi polutan dapat merepresentasikan kondisi kualitas udara yang sebenarnya.

### Cleaning Data

#### Fixing ... problem

```python
# Membuat kolom datetime dari year, month, day, dan hour
df['datetime'] = pd.to_datetime(
    df[['year', 'month', 'day', 'hour']]
)

# Mengurutkan data berdasarkan stasiun dan waktu
df = df.sort_values(['station', 'datetime']).reset_index(drop=True)

# Melihat hasilnya
df[['station', 'datetime']].head()
```

```python
# Kolom numerik yang akan ditangani missing value-nya
kolom_numerik = [
    'PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3',
    'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM'
]

# Interpolasi berdasarkan waktu untuk setiap stasiun
df[kolom_numerik] = (
    df.groupby('station', group_keys=False)[kolom_numerik]
      .apply(lambda group: group.interpolate(method='linear', limit_direction='both'))
)
```

```python
# Mengisi missing value pada arah angin (wd)
df['wd'] = df.groupby('station')['wd'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
```

```python
# Mengecek kembali missing value
missing_after_cleaning = df.isna().sum()

missing_after_cleaning[missing_after_cleaning > 0]
```

**Insight:** (Opsional)
- Missing value pada variabel numerik ditangani menggunakan interpolasi linear berdasarkan urutan waktu pada masing-masing stasiun.
- Missing value pada variabel arah angin (`wd`) ditangani menggunakan nilai modus pada masing-masing stasiun.
- Outlier tidak dihapus karena nilai ekstrem pada konsentrasi polutan dapat merepresentasikan kondisi kualitas udara yang sebenarnya. Outlier akan tetap dipertahankan dan dipertimbangkan dalam interpretasi hasil analisis.

## Exploratory Data Analysis (EDA)

### Explore 1 Perubahan rata-rata PM2.5 dan CO berdasarkan tahun

### **Pertanyaan 1**
Bagaimana perubahan rata-rata konsentrasi PM2.5 dan CO berdasarkan tahun selama periode 2013–2017, dan tahun mana yang memiliki tingkat polusi tertinggi?


```python
q1 = df.groupby('year')[['PM2.5', 'CO']].mean()

print("=== Q1: Rata-rata PM2.5 dan CO Berdasarkan Tahun ===")
print(q1)

# Tahun dengan rata-rata PM2.5 tertinggi
tahun_pm25_tertinggi = q1['PM2.5'].idxmax()
nilai_pm25_tertinggi = q1['PM2.5'].max()

# Tahun dengan rata-rata CO tertinggi
tahun_co_tertinggi = q1['CO'].idxmax()
nilai_co_tertinggi = q1['CO'].max()

print("\nTahun dengan rata-rata PM2.5 tertinggi:",
      tahun_pm25_tertinggi)
print("Nilai rata-rata PM2.5:",
      nilai_pm25_tertinggi)

print("\nTahun dengan rata-rata CO tertinggi:",
      tahun_co_tertinggi)
print("Nilai rata-rata CO:",
      nilai_co_tertinggi)
```

### Explore 2 Rata-rata PM2.5 berdasarkan stasiun

### **Pertanyaan 2**
Stasiun pemantauan mana yang memiliki rata-rata konsentrasi PM2.5 tertinggi selama periode 2013–2017, dan bagaimana kondisi cuaca seperti temperatur, tekanan udara, kelembapan titik embun, curah hujan, dan kecepatan angin berkaitan dengan konsentrasi PM2.5?

```python
q2_stasiun = (
    df.groupby('station')['PM2.5']
      .mean()
      .sort_values(ascending=False)
)

print("=== Rata-rata PM2.5 Berdasarkan Stasiun ===")
print(q2_stasiun)


# Mencari stasiun dengan rata-rata PM2.5 tertinggi
stasiun_pm25_tertinggi = q2_stasiun.idxmax()
nilai_pm25_stasiun_tertinggi = q2_stasiun.max()

print("\n=== Stasiun dengan Rata-rata PM2.5 Tertinggi ===")
print("Stasiun:", stasiun_pm25_tertinggi)
print("Rata-rata PM2.5:", nilai_pm25_stasiun_tertinggi)


# Korelasi PM2.5 dengan faktor cuaca
q2_korelasi = df[
    ['PM2.5', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']
].corr()['PM2.5'].sort_values(ascending=False)

print("\n=== Korelasi PM2.5 dengan Faktor Cuaca ===")
print(q2_korelasi)
```

**Insight:** (Opsional)
- Perubahan konsentrasi PM2.5 dan CO berbeda antarperiode tahun.
- Konsentrasi PM2.5 juga berbeda antarstasiun dan memiliki hubungan dengan beberapa kondisi cuaca.

## Analisis Lanjutan (Opsional)

```python
import matplotlib.pyplot as plt

# Faktor cuaca yang dianalisis
faktor_cuaca = ['TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']

# Menghitung korelasi PM2.5 dengan faktor cuaca
korelasi_pm25 = (
    df[['PM2.5'] + faktor_cuaca]
    .corr()['PM2.5']
    .drop('PM2.5')
    .sort_values()
)

# Menampilkan hasil korelasi
print("=== Korelasi PM2.5 dengan Faktor Cuaca ===")
print(korelasi_pm25)

# Membuat visualisasi korelasi
plt.figure(figsize=(10, 5))

korelasi_pm25.plot(kind='bar')

plt.title('Korelasi PM2.5 dengan Faktor Cuaca')
plt.xlabel('Faktor Cuaca')
plt.ylabel('Koefisien Korelasi')
plt.axhline(0, linewidth=1)
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

**Insight:** (Opsional)

- Analisis lanjutan menunjukkan adanya hubungan antara konsentrasi PM2.5 dengan beberapa faktor cuaca.
- Faktor cuaca dengan nilai korelasi absolut terbesar memiliki hubungan paling kuat dengan konsentrasi PM2.5 selama periode pengamatan.
- Hasil analisis ini dapat menjadi bahan pertimbangan untuk memahami kondisi lingkungan yang berkaitan dengan perubahan konsentrasi PM2.5.
